```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.mean()))
```

```python
# Convert column names to lowercase
df.columns = df.columns.str.lower()
```

```python
# Replace spaces with underscores
df.columns = df.columns.str.replace(' ', '_')
```

```python
# Rename purchase amount column
df = df.rename(columns={
    'purchase_amount_(usd)':'purchase_amount'
})
```

Age group


```python
labels = ['young adults', 'adults', 'middle-aged', 'seniors']
df['age_group'] = pd.qcut(df['age'], q=4, labels =labels)
```

```python
df[['age','age_group']].head(10)
```

Create coloum purchase_ferquency_days

```python
frequency_mapping ={
    'fortnightly': 14,
    'weekly': 7,
    'monthly': 30,  
    'Quarterly': 90,
    'Bi-weekly': 14,
    'annually': 365,
    'Every 3 Months': 90,

}
df['previous_purchases_days'] = df['previous_purchases'].map(frequency_mapping)
```

```python
from sqlalchemy import create_engine, text
from urllib.parse import quote

# MySQL connection
username = "root"
password = "Reddy@123"
host = "localhost"
port = "3306"
database = "customer_behavior"

# Encode password if it contains special characters like @
password_encoded = quote(password, safe="")

engine = create_engine(
    f"mysql+pymysql://{username}:{password_encoded}@{host}:{port}/{database}"
)

# Write DataFrame to MySQL
table_name = "customers"

df.to_sql(
    table_name,
    engine,
    if_exists="replace",
    index=False
)

# Read back sample data
query = "SELECT * FROM customers LIMIT 5"

df_read = pd.read_sql(query, engine)

print(df_read)
```
